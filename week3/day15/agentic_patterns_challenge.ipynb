### Task 3.1: Router Agent (Routing Pattern)

**Requirements:**
- Create a router that classifies content into: "social_media", "article", or "comment"
- Route should be based on length, structure, and style
- Return the classification decision

````python
# TODO: Implement Router Agent
# This agent analyzes content and classifies it

def router_agent(content):
    words = content.strip().split()
    word_count = len(words)
    if word_count < 15:
        return "comment"
    elif word_count < 60:
        if "#" in content or "@" in content or "!" in content or content.lower().count("i ") > 1:
            return "social_media"
        else:
            return "article"
    else:
        return "article"

content = "just finished reading an amzing book about AI ethics! its really make me think about how we build responsible systems."
print("Router Agent Output:", router_agent(content))
````

### Task 3.2: Parallel Analysis Agents (Parallelization Pattern)

**Requirements:**
- Implement 3 agents that run in parallel:
  1. **Safety Checker**: Detect toxic, harmful, or inappropriate content
  2. **Tone Analyzer**: Assess sentiment (positive/negative/neutral) and professionalism
  3. **Grammar Checker**: Identify spelling, grammar, and clarity issues
- Each agent should return a structured assessment
- Execute them concurrently for efficiency

````python
# TODO: Implement Parallel Analysis Agents
# These agents analyze different aspects simultaneously

def safety_checker(content):
    harmful_words = ["hate", "stupid", "idiot", "terrible", "waste"]
    for word in harmful_words:
        if word in content.lower():
            return {"safe": False, "reason": f"Harmful word detected: {word}"}
    return {"safe": True, "reason": "No harmful language found"}

def tone_analyzer(content):
    positive_words = ["good", "great", "amazing", "love", "recommend"]
    negative_words = ["hate", "bad", "terrible", "waste", "stupid"]
    score = 0
    for word in positive_words:
        if word in content.lower():
            score += 1
    for word in negative_words:
        if word in content.lower():
            score -= 1
    if score > 0:
        sentiment = "positive"
    elif score < 0:
        sentiment = "negative"
    else:
        sentiment = "neutral"
    return {"sentiment": sentiment, "score": score}

def grammar_checker(content):
    mistakes = 0
    if "ur" in content.lower():
        mistakes += 1
    if "grate" in content.lower():
        mistakes += 1
    if "amzing" in content.lower():
        mistakes += 1
    if "recomend" in content.lower():
        mistakes += 1
    return {"mistakes": mistakes, "issues": "Basic spelling/grammar issues" if mistakes > 0 else "No major issues"}

def run_parallel_agents(content):
    return {
        "safety": safety_checker(content),
        "tone": tone_analyzer(content),
        "grammar": grammar_checker(content)
    }

analysis_results = run_parallel_agents(content)
print("Parallel Analysis Output:", analysis_results)
````

### Task 3.3: Evaluator Agent (Evaluator-Optimizer Pattern - Part 1)

**Requirements:**
- Aggregate results from the 3 parallel agents
- Calculate an overall content quality score (0-100)
- Make a decision: APPROVE (score ≥ 70) or REJECT (score < 70)
- For approved content, provide specific improvement suggestions

````python
# TODO: Implement Evaluator Agent
# This agent aggregates findings and makes decisions

def evaluator_agent(analysis_results):
    score = 100
    if not analysis_results["safety"]["safe"]:
        score -= 50
    if analysis_results["tone"]["sentiment"] == "negative":
        score -= 20
    elif analysis_results["tone"]["sentiment"] == "neutral":
        score -= 5
    score -= analysis_results["grammar"]["mistakes"] * 5
    score = max(0, min(100, score))
    if score >= 70:
        decision = "APPROVE"
        suggestions = []
        if analysis_results["grammar"]["mistakes"] > 0:
            suggestions.append("Fix grammar and spelling mistakes.")
        if analysis_results["tone"]["sentiment"] == "neutral":
            suggestions.append("Try to make the tone more engaging.")
        if not suggestions:
            suggestions.append("Content is good, just minor improvements possible.")
    else:
        decision = "REJECT"
        suggestions = []
        if not analysis_results["safety"]["safe"]:
            suggestions.append("Remove harmful or toxic language.")
        if analysis_results["tone"]["sentiment"] == "negative":
            suggestions.append("Make the tone more positive or constructive.")
        if analysis_results["grammar"]["mistakes"] > 0:
            suggestions.append("Fix grammar and spelling mistakes.")
    return {
        "score": score,
        "decision": decision,
        "suggestions": suggestions
    }

evaluation = evaluator_agent(analysis_results)
print("Evaluator Output:", evaluation)
````

### Task 3.4: Optimizer & Enhancer Agents (Prompt Chaining + Evaluator-Optimizer)

**Requirements:**
- **Optimizer Agent**: Generate specific improvements based on evaluator feedback
- **Enhancer Agent**: Apply improvements to create an enhanced version
- Implement as a chain: Original Content → Optimizer → Enhancer → Final Content
- (Optional) Add a re-evaluation loop if initial enhancement score is still low

````python
# TODO: Implement Optimizer and Enhancer Agents
# These agents improve content based on feedback

def optimizer_agent(content, suggestions):
    improved = content
    for suggestion in suggestions:
        if "grammar" in suggestion or "spelling" in suggestion:
            improved = improved.replace("amzing", "amazing").replace("recomend", "recommend")
        if "engaging" in suggestion:
            improved += " I highly recommend it to everyone interested in tech!"
    return improved

def enhancer_agent(improved_content):
    sentences = improved_content.split(". ")
    enhanced = ". ".join(s.capitalize() for s in sentences)
    if not enhanced.endswith("."):
        enhanced += "."
    return enhanced

optimized = optimizer_agent(content, evaluation["suggestions"])
print("Optimizer Output:", optimized)

enhanced = enhancer_agent(optimized)
print("Enhancer Output:", enhanced)
````

### Task 3.5: Orchestrator (Orchestrator-Worker Pattern)

**Requirements:**
- Create a master orchestrator that coordinates the entire pipeline:
  1. Route content type
  2. Run parallel analysis
  3. Evaluate and decide
  4. If approved, optimize and enhance
  5. Return final result with metadata
- Handle both approval and rejection cases
- Provide clear logging of each step

````python
# TODO: Implement Orchestrator
# This coordinates the entire moderation pipeline

def orchestrator(content):
    print("Step 1: Routing...")
    content_type = router_agent(content)
    print("Content Type:", content_type)

    print("Step 2: Parallel Analysis...")
    analysis = run_parallel_agents(content)
    print("Analysis Results:", analysis)

    print("Step 3: Evaluation...")
    evaluation = evaluator_agent(analysis)
    print("Evaluation:", evaluation)

    result = {
        "content_type": content_type,
        "analysis": analysis,
        "evaluation": evaluation
    }

    if evaluation["decision"] == "APPROVE":
        print("Step 4: Optimization & Enhancement...")
        optimized = optimizer_agent(content, evaluation["suggestions"])
        print("Optimizer Output:", optimized)
        enhanced = enhancer_agent(optimized)
        print("Enhancer Output:", enhanced)
        result["final_content"] = enhanced
    else:
        print("Content was rejected. No enhancement applied.")
        result["final_content"] = None

    return result

test_content = "just finished reading an amzing book about AI ethics! its really make me think about how we build responsible systems."
final_result = orchestrator(test_content)
````

**1. Routing Pattern:**

- Why effective: All three agents check the content at the same time, so it is faster.
- Performance benefit: It saves time because the checks do not wait for each other.

**3. Evaluator-Optimizer Pattern:**
- Where used:  I used it in the evaluator_agent function and the optimizer_agent function.
- How feedback loop works: The evaluator gives suggestions and the optimizer uses these suggestions to improve the content.

**4. Prompt Chaining Pattern:**
- Where used: I used it in the chain from optimizer_agent to enhancer_agent.
- Stages in chain:

Original Content

Optimizer Agent

Enhancer Agent

Final Enhanced Content

**5. Orchestrator-Worker Pattern:**
- How orchestration is managed: The orchestrator function manages all steps in order.
- Worker coordination:  Each agent does its job and the orchestrator collects and uses their results.
